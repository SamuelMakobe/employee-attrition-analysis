# Part-3 Initial Data Exploration

```python
import matplotlib.pyplot as plt   #importing the matplotlib for creating visualisations
```

# 1.Employee attrition distribution

```python
attrition_counts = df["Attrition"].value_counts()          #Counting the number of employees in each attrition (who stayed and employees who left the company)

attrition_counts.plot(kind="bar")                          #Creating the bar graph

plt.title("Employee attrition distribution")               #Adding the title to the created graph (Employee attrition distribution)

plt.xlabel("Attrition")                                    #Adding the x-axis label (Attrition)

plt.ylabel("Number of Employees")                          #Adding the y-axis labe (Number of employees)

plt.show()                                                 #Displaying the graph wth everything included (title,x-axis and y-axis label)
```

# Interpretation: The bar chart shows that most employees in the dataset remained with the company, while a smaller proportion left. Out of 1,470 employees, 1,233 (83.88%) did not leave, whereas 237 (16.12%) experienced attrition. This indicates that employee attrition affects a minority of the workforce, but the 16.12% attrition rate is still significant enough to warrant further investigation into the factors associated with employee turnover.

# 2.Attrition by department

```python
department_attrition=pd.crosstab(df["Department"],df["Attrition"])      #Creating a cross-tabulation of department and attrition

department_attrition.plot(kind="bar")                                   #creating a bar graph

plt.title("Employee attrition by department")                           #adding a title to a graph (Employee attrition by department)

plt.xlabel("Department")                                                #adding the x-axis label (Department)

plt.ylabel("Number of employees")                                       #adding the y-axis labe(Number of employees)

plt.show()                                                              #Displaying the created graph with all variables
```

# Analysis for chart 2: Attrition by department

```python
department_attrition_rate = pd.crosstab(                                #Calculating the percentage of employees who left within each department
    df["Department"],
    df["Attrition"],
    normalize="index"
) * 100
      
department_attrition_rate                                               #Displaying the attrition rates for each department
```

# Interpretation: The attrition rate differs across the three departments. The Sales department has the highest attrition rate at 20.63%, followed by Human Resources at 19.05%. Research & Development has the lowest attrition rate at 13.84%. This suggests that employee attrition is more prevalent in Sales and Human Resources than in Research & Development. However, these results show an association rather than a cause of employee attrition, and further analysis is required to identify the factors contributing to these differences.

# 3.Attrition by job role

```python
jobrole_attrition=pd.crosstab(df["JobRole"],df["Attrition"])     #Creating a cross-tabulation of JobRole and attrition

jobrole_attrition.plot(kind="bar")                               #creating a bar graph

plt.title("Employee attrition by job role")                      #adding a title to a graph (Employee attrition by job role)

plt.xlabel("Job role")                                           #adding the x-axis label (Job role)

plt.ylabel ("Number of employees")                               #adding the y-axis label (Number of employees)

plt.show()                                                       #Displaying the created graph with all variables
```

# Analysis for chart 3: Attrition by job role

```python
jobrole_attrition_rate = pd.crosstab(                                    #Calculating the percentage of employees who left within each job role
    df["JobRole"], 
    df["Attrition"],
    normalize="index"
) * 100

jobrole_attrition_rate                                                   #Displaying the attrition rates for each job role
```

# Interpretation: The attrition rate varies considerably across job roles. Sales Representatives have the highest attrition rate at 39.76%, followed by Laboratory Technicians at 23.94% and Human Resources employees at 23.08%. In contrast, Research Directors have the lowest attrition rate at 2.50%, followed by Managers at 4.90%. The substantially higher attrition rate among Sales Representatives suggests that job role may be an important factor to investigate further when examining employee turnover. However, these results indicate an association and do not establish that job role directly causes employee attrition.

# 4.1 Workforce observations

1.	Overall employee attrition

This indicates that although most employees remained, a notable proportion of the workforce left during the period represented by the dataset.

2.	Variation in attrition across departments
  
Employee attrition differs across departments. The Sales department recorded the highest attrition rate at 20.63%, followed by Human Resources at 19.05%. In comparison, the Research & Development department had the lowest attrition rate at 13.84%. These differences suggest that employees' likelihood of leaving may vary depending on the department in which they work.

3.	Variation in attrition across job roles
  
A considerable difference in attrition can also be observed across job roles. Sales Representatives had the highest attrition rate at 39.76%, whereas Research Directors had the lowest rate at 2.50%. The large difference between these roles suggests that certain job specific factors may be associated with employee turnover and should be explored further.

4.	Age differences between employees who stayed and those who left
  
There is also a noticeable variation in the average age of employees who stayed compared with those who left the company. Employees who left the company had an average age of 33.61 years, while those who remained had an average age of 37.56 years. This suggests that younger employees may be more likely to leave the organization within this dataset.

5.	Higher attrition among employees working overtime

Overtime appears to be linked with a substantially higher level of employee attrition. Employees who worked overtime had an attrition rate of 30.53%, compared with 10.44% among those who did not work overtime. The considerable difference between these groups makes overtime an important factor for further investigation.


# 4.2 Possible factors contributing to employee attrition

1.	Job role
  
Job role appears to be a significant factor to consider when evaluating employee attrition. Sales Representatives had an attrition rate of 39.76%, which was considerably higher than that of most other job roles. This may indicate that the responsibilities, working conditions, workload, or career opportunities associated with particular roles could be related to employee turnover. However, further analysis would be necessary to understand the reasons behind this difference.

2.	Age
  
Age may also be related to employee attrition. Employees who left the organization had an average age of 33.61 years, compared with 37.56 years among employees who stayed. This difference suggests that younger employees may have a higher likelihood of leaving in this dataset. Further investigation could help determine whether age itself is important or whether it reflects other factors, such as job level, income, or years of experience.

3.	Monthly income
  
Monthly income may also play a vital role in employee retention. The boxplot indicated differences in the income distributions between employees who stayed and those who left the company. This suggests that compensation may be worth examining further as a potential factor associated with employee turnover. Additional analysis would be needed to determine whether lower income levels are consistently associated with higher attrition.

4.	Overtime
  
Overtime stands out as one of the factors that warrants further investigation. Employees who worked overtime had an attrition rate of 30.53%, compared with 10.44% among employees who did not work overtime. This substantial difference suggests that employees who regularly work overtime may be more likely to leave the organization. However, the analysis only identifies an association and does not establish that overtime directly causes employees to leave.

5.	Job satisfaction
  
Job satisfaction also appears to be related to employee attrition. Employees with the lowest job satisfaction level had an attrition rate of 22.84%, compared with 11.33% among employees with the highest satisfaction level. This suggests that lower levels of job satisfaction may be associated with a greater likelihood of employees leaving the company. Understanding the factors that influence job satisfaction could therefore be useful when developing strategies to improve employee retention.
Overall Business Insight

Overall, the initial analysis suggests that employee attrition is not evenly distributed across the workforce. Noticeable differences can be observed based on department, job role, age, overtime status, monthly income, and job satisfaction. In particular, the relatively high attrition rate among Sales Representatives and employees who work overtime highlights areas that may require closer investigation. These findings provide useful initial insights into patterns within the workforce; however, they should be interpreted with caution. The analysis identifies associations rather than direct causes of employee attrition. For example, the higher attrition observed among employees working overtime does not necessarily mean that overtime alone causes employees to leave.
